# Claude 3 Haiku로 웹 페이지 내용 요약하기
이 레시피에서는 URL로 웹 페이지의 내용을 가져온 뒤, Anthropic Claude API로 그 내용을 요약하는 방법을 알아봅니다.

먼저 Anthropic 라이브러리를 설치하겠습니다.

## 준비
먼저 필요한 라이브러리를 설치하고, API 키로 Anthropic 클라이언트를 설정합니다.

```python
# Install the necessary libraries
%pip install anthropic
```

```python
# Import the required libraries
from anthropic import Anthropic

# Set up the Claude API client
client = Anthropic()
MODEL_NAME = "claude-haiku-4-5"
```

## 1단계: 웹 페이지 내용 가져오기
먼저 주어진 URL로 웹 페이지의 내용을 가져와야 합니다. 이를 위해 requests 라이브러리를 사용합니다.

```python
import requests

url = "https://en.wikipedia.org/wiki/96th_Academy_Awards"
response = requests.get(url, timeout=30)

if response.status_code == 200:
    page_content = response.text
else:
    print(f"Failed to fetch the web page. Status code: {response.status_code}")
    exit(1)
```

## 2단계: Claude에 보낼 입력 준비하기
다음으로 Claude API에 보낼 입력을 준비합니다. 페이지 내용과 함께 요약을 요청하는 프롬프트를 담은 메시지를 만듭니다.

```python
prompt = (
    f"<content>{page_content}</content>Please produce a concise summary of the web page content."
)

messages = [{"role": "user", "content": prompt}]
```

## 3단계: 요약 생성하기
이제 Haiku를 호출해 웹 페이지 내용의 요약을 생성합니다.

```python
response = client.messages.create(model="claude-haiku-4-5", max_tokens=1024, messages=messages)

summary = response.content[0].text
print(summary)
```
